# **Problem Statement**

## **Business Context**

Workplace safety in hazardous environments like construction sites and industrial plants is crucial to prevent accidents and injuries. One of the most important safety measures is ensuring workers wear safety helmets, which protect against head injuries from falling objects and machinery. Non-compliance with helmet regulations increases the risk of serious injuries or fatalities, making effective monitoring essential, especially in large-scale operations where manual oversight is prone to errors and inefficiency.

To overcome these challenges, SafeGuard Corp plans to develop an automated image analysis system capable of detecting whether workers are wearing safety helmets. This system will improve safety enforcement, ensuring compliance and reducing the risk of head injuries. By automating helmet monitoring, SafeGuard aims to enhance efficiency, scalability, and accuracy, ultimately fostering a safer work environment while minimizing human error in safety oversight.

## **Objective**

As a data scientist at SafeGuard Corp, you are tasked with developing an image classification model that classifies images into one of two categories:
- **With Helmet:** Workers wearing safety helmets.
- **Without Helmet:** Workers not wearing safety helmets.

## **Data Description**

The dataset consists of **4125 images**, divided into two categories:

- **With Helmet:** 3161 images showing workers wearing helmets.
- **Without Helmet:** 964 images showing workers not wearing helmets.

**Dataset Characteristics:**
- **Variations in Conditions:** Images include diverse environments such as construction sites, factories, and industrial settings, with variations in lighting, angles, and worker postures to simulate real-world conditions.
- **Worker Activities:** Workers are depicted in different actions such as standing, using tools, or moving, ensuring robust model learning for various scenarios.

###Plot random images from each of the classes and print their corresponding labels.

```python
def plot_random_images_from_class(images_dataset, labels_df, class_label, num_images):
    """
    Plots a grid of random images from a specified class.

    Args:
        images_dataset (np.ndarray): The dataset of images.
        labels_df (pd.DataFrame): DataFrame containing image labels.
        class_label (int): The label of the class to display (0 for 'Without Helmet', 1 for 'With Helmet').
        num_images (int): The number of random images to display.
    """
    # Filter indices for the specified class
    class_indices = labels_df[labels_df['label'] == class_label].index

    # Randomly select num_images indices
    selected_indices = np.random.choice(class_indices, num_images, replace=False)

    # Determine grid size for subplots
    cols = 4  # Number of columns for the grid
    rows = math.ceil(num_images / cols)

    plt.figure(figsize=(cols * 4, rows * 4))
    for i, idx in enumerate(selected_indices):
        plt.subplot(rows, cols, i + 1)
        plt.imshow(images_dataset[idx])
        plt.title(f"Class: {class_label}")
        plt.axis('off')
    plt.tight_layout()
    plt.show()

# Call the function for 'Without Helmet' (class 0)
print("Random images for 'Without Helmet':")
plot_random_images_from_class(images, labels, class_label=0, num_images=8)

# Call the function for 'With Helmet' (class 1)
print("\nRandom images for 'With Helmet':")
plot_random_images_from_class(images, labels, class_label=1, num_images=8)
```

## RGB Pixel Intensity Distributions Per Class

```python
classes = {0: 'Without Helmet', 1: 'With Helmet'}
colors = {'R': 'red', 'G': 'green', 'B': 'blue'}

# Step 1: Initialize global_max_density
global_max_density = 0

# Step 2-4: First loop to calculate global_max_density
for class_label in classes.keys():
    # Filter images for the current class
    class_images = images[labels[labels['label'] == class_label].index]

    # Iterate through each channel (R, G, B)
    for channel_idx in range(3):
        # Flatten the pixel values for the current channel
        channel_pixels = class_images[:, :, :, channel_idx].flatten()

        # Calculate histogram densities (hist) and bin edges (bins)
        hist, _ = np.histogram(channel_pixels, bins=256, range=(0, 256), density=True)

        # Find the maximum density in the current histogram
        max_current_density = hist.max()

        # Update global_max_density if a higher density is found
        if max_current_density > global_max_density:
            global_max_density = max_current_density

# Step 5-8: Second loop to plot histograms with consistent y-axis
for class_label, class_name in classes.items():
    # Filter images for the current class
    class_images = images[labels[labels['label'] == class_label].index]

    # Directly flatten pixel values for each channel using NumPy
    r_pixels = class_images[:, :, :, 0].flatten()
    g_pixels = class_images[:, :, :, 1].flatten()
    b_pixels = class_images[:, :, :, 2].flatten()

    plt.figure(figsize=(18, 5))
    plt.suptitle(f'Pixel Intensity Histograms for {class_name}', fontsize=16)

    # Plot histogram for Red channel
    plt.subplot(1, 3, 1)
    plt.hist(r_pixels, bins=256, color=colors['R'], alpha=0.7, range=(0, 256), density=True)
    plt.title('Red Channel')
    plt.xlabel('Pixel Intensity')
    plt.ylabel('Density')
    plt.xlim([0, 256])
    plt.ylim(0, global_max_density) # Apply global max density

    # Plot histogram for Green channel
    plt.subplot(1, 3, 2)
    plt.hist(g_pixels, bins=256, color=colors['G'], alpha=0.7, range=(0, 256), density=True)
    plt.title('Green Channel')
    plt.xlabel('Pixel Intensity')
    plt.ylabel('Density')
    plt.xlim([0, 256])
    plt.ylim(0, global_max_density) # Apply global max density

    # Plot histogram for Blue channel
    plt.subplot(1, 3, 3)
    plt.hist(b_pixels, bins=256, color=colors['B'], alpha=0.7, range=(0, 256), density=True)
    plt.title('Blue Channel')
    plt.xlabel('Pixel Intensity')
    plt.ylabel('Density')
    plt.xlim([0, 256])
    plt.ylim(0, global_max_density) # Apply global max density

    plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout to prevent suptitle overlap
    plt.show()
```

## Utility Functions

```python
# defining a function to compute different metrics to check performance of a classification model built using statsmodels
def model_performance_classification(model, predictors, target):
    """
    Function to compute different metrics to check classification model performance

    model: classifier
    predictors: independent variables
    target: dependent variable
    """

    # checking which probabilities are greater than threshold
    pred = model.predict(predictors).reshape(-1)>0.5

    target = target.to_numpy().reshape(-1)


    acc = accuracy_score(target, pred)  # to compute Accuracy
    recall = recall_score(target, pred, average='micro')  # to compute Recall
    precision = precision_score(target, pred, average='micro')  # to compute Precision
    f1 = f1_score(target, pred, average='micro')  # to compute F1-score

    # creating a dataframe of metrics
    df_perf = pd.DataFrame({"Accuracy": acc, "Recall": recall, "Precision": precision, "F1 Score": f1,},index=[0],)

    return df_perf
```

```python
def plot_confusion_matrix(model,predictors,target,ml=False):
    """
    Function to plot the confusion matrix

    model: classifier
    predictors: independent variables
    target: dependent variable
    ml: To specify if the model used is an sklearn ML model or not (True means ML model)
    """

    # checking which probabilities are greater than threshold
    pred = model.predict(predictors).reshape(-1)>0.5

    target = target.to_numpy().reshape(-1)

    # Plotting the Confusion Matrix using confusion matrix() function which is also predefined tensorflow module
    confusion_matrix = tf.math.confusion_matrix(target,pred)
    f, ax = plt.subplots(figsize=(10, 8))
    sns.heatmap(
        confusion_matrix,
        annot=True,
        linewidths=.4,
        fmt="d",
        square=True,
        ax=ax
    )
    plt.show()
```

```python
def plot(history, name):
    """
    Function to plot loss/accuracy

    history: an object which stores the metrics and losses.
    name: can be one of Loss or Accuracy
    """
    fig, ax = plt.subplots() #Creating a subplot with figure and axes.
    plt.plot(history.history[name]) #Plotting the train accuracy or train loss
    plt.plot(history.history['val_'+name]) #Plotting the validation accuracy or validation loss

    plt.title('Model ' + name.capitalize()) #Defining the title of the plot.
    plt.ylabel(name.capitalize()) #Capitalizing the first letter.
    plt.xlabel('Epoch') #Defining the label for the x-axis.
    fig.legend(['Train', 'Validation'], loc="outside right upper") #Defining the legend, loc controls the position of the legend.
```

```python
def plot_loss_recall_accuracy_precision(history):
  plot(history, 'loss')
  plot(history, 'recall')
  plot(history, 'accuracy')
  plot(history, 'precision')
```

```python
def train_with_early_stopping(model, x_train_data, y_train_data, x_val_data, y_val_data, batch_size, initial_history=None, patience=10, initial_epochs=0, verbose=1, class_weights=None):
    best_val_loss = float('inf')
    epochs_without_improvement = 0
    total_epochs_trained = initial_epochs
    best_model_weights = None
    best_epoch_metrics = None # Store the best metrics, not just the loss

    # Initialize local history variable based on initial_history parameter
    history = initial_history
    # so that we can create an updated history without using the misleading variable name "initial_history" everywhere

    if history is None or not history.history or not history.history['val_loss']:
        print("Starting new training session with no prior history.")
    else:
        best_val_loss = history.history['val_loss'][-1]
        total_epochs_trained = initial_epochs + len(history.history['loss'])
        print(f"Continuing training from initial validation loss: {best_val_loss:.4f}")
        # If continuing, initialize best_epoch_metrics with the last recorded metrics
        best_epoch_metrics = {key: history.history[key][-1] for key in history.history}

    start_time_overall = time.time()

    # Determine if x_train_data is an ImageDataGenerator.Iterator
    is_generator = isinstance(x_train_data, (tf.keras.preprocessing.image.ImageDataGenerator, tf.keras.preprocessing.image.Iterator))

    while True:
        if is_generator:
            # When using a generator, y_train_data should be None and steps_per_epoch is required
            steps_per_epoch_val = math.ceil(x_train_data.n / batch_size) # Corrected from x_train_data.labels to x_train_data.n
            new_history = model.fit(
                x_train_data, # Generator directly passed
                epochs=1,  # Train for one epoch at a time
                steps_per_epoch=steps_per_epoch_val, # Use steps_per_epoch for generators
                validation_data=(x_val_data, y_val_data),
                verbose=verbose,
                class_weight=class_weights if class_weights is not None else None
            )
        else:
            new_history = model.fit(
                x_train_data, y_train_data,
                epochs=1,  # Train for one epoch at a time
                batch_size=batch_size,
                validation_data=(x_val_data, y_val_data),
                verbose=verbose,
                class_weight=class_weights if class_weights is not None else None
            )
        total_epochs_trained += 1

        current_val_loss = new_history.history['val_loss'][-1]
        print(f"Epoch {total_epochs_trained}: Validation loss: {current_val_loss:.4f}")

        # Combine the new history with the overall history
        if history is None:
            history = new_history
        else:
            for key in new_history.history:
                if key in history.history:
                    history.history[key].extend(new_history.history[key])
                else:
                    history.history[key] = new_history.history[key]

        if current_val_loss < best_val_loss:
            best_val_loss = current_val_loss
            epochs_without_improvement = 0
            best_model_weights = model.get_weights() # Save the weights of the best model
            best_epoch_metrics = {key: new_history.history[key][-1] for key in new_history.history} # Save metrics for this best epoch
            print("Validation loss improved. Model weights saved.")
        else:
            epochs_without_improvement += 1
            print(f"Validation loss did not improve. Epochs without improvement: {epochs_without_improvement}")
        print('\n')

        if epochs_without_improvement >= patience:
            print(f"\nEarly stopping triggered after {patience} epochs without improvement.")
            break

    end_time_overall = time.time()
    total_time_trained = end_time_overall - start_time_overall

    # Load best weights if they were saved
    if best_model_weights is not None:
        model.set_weights(best_model_weights);
        print("Loaded best model weights.")
    else:
        # If no improvement was detected, the best model is the one from the last epoch
        # and its metrics are already in the history. We just need to ensure best_epoch_metrics is set.
        if best_epoch_metrics is None and history is not None:
             best_epoch_metrics = {key: history.history[key][-1] for key in history.history}
        print("No improvement in validation loss detected during training. Returning model from last epoch.")

    return model, history, total_epochs_trained, total_time_trained, best_epoch_metrics
```

```python
def new_results_entry(model, history, total_epochs, best_epoch_metrics, time, x_train_data, y_train_data, x_val_data, y_val_data, batch_size=64):

  # 1. Number of Hidden Layers (all layers except the last one, which is the output layer)
  num_hidden_layers = len(model.layers) - 1

  # 2. Hidden Layer Sizes (units of all layers except the last one)
  hidden_layer_sizes = [layer.units for layer in model.layers[:-1] if isinstance(layer, Dense)]

  # 3. Activation Function Names (for all layers except the last one)
  activation_function_names = [layer.activation.__name__ for layer in model.layers[:-1] if isinstance(layer, Dense)]

  # 4. Total Epochs & 5. Batch Size

  # 6. Optimizer Name and 7. Learning Rate/Momentum
  optimizer_instance = model.optimizer
  optimizer_name = optimizer_instance.__class__.__name__ # Get class name (e.g., 'Adam', 'SGD')

  learning_rate_value = None
  if hasattr(optimizer_instance, 'learning_rate'):
      lr_obj = optimizer_instance.learning_rate
      if callable(lr_obj):
          learning_rate_value = lr_obj()
      elif isinstance(lr_obj, (tf.Tensor, tf.Variable)):
          learning_rate_value = lr_obj.numpy()
      else:
          learning_rate_value = lr_obj

  learning_rate_rounded = round(float(learning_rate_value), 4) if learning_rate_value is not None else '-'

  momentum_value = 0.0 # Default to 0, only for SGD
  if isinstance(optimizer_instance, keras.optimizers.SGD):
      if hasattr(optimizer_instance, 'momentum'):
          mom_obj = optimizer_instance.momentum
          if callable(mom_obj):
              momentum_value = mom_obj()
          elif isinstance(mom_obj, (tf.Tensor, tf.Variable)):
              momentum_value = mom_obj.numpy()
          else:
              momentum_value = mom_obj

  momentum_rounded = round(float(momentum_value), 4) if isinstance(momentum_value, (int, float)) else 0.0

  # 8. Weight Initializer Name
  weight_initializer_name = 'Default (GlorotUniform)' # Most common default in Keras
  if hasattr(model.layers[0], 'kernel_initializer'):
      weight_initializer_name = model.layers[0].kernel_initializer.__class__.__name__

  # 9. Regularization
  regularization_name = '-'
  if hasattr(model.layers[0], 'kernel_regularizer'):
      if model.layers[0].kernel_regularizer is not None:
        regularization_name = model.layers[0].kernel_regularizer.__class__.__name__

  # 10. Dropout Rates
  dropout_rates = [layer.rate for layer in model.layers if isinstance(layer, Dropout)]

  # 11. - 15. Train Loss, Accuracy, Recall, Precision, & F1
  # Retrieve training metrics directly from best_epoch_metrics
  train_loss = best_epoch_metrics['loss']
  train_accuracy = best_epoch_metrics['accuracy']
  train_recall = best_epoch_metrics['recall']
  train_precision = best_epoch_metrics['precision']

  # Calculate F1 score, handling potential division by zero
  if (train_precision + train_recall) == 0:
      train_f1 = 0.0
  else:
      train_f1 = 2 * (train_precision * train_recall) / (train_precision + train_recall)

  # 16. - 20. Validation Loss, Accuracy, Recall, Precision, & F1
  val_loss = best_epoch_metrics['val_loss']
  val_accuracy = best_epoch_metrics['val_accuracy']
  val_recall = best_epoch_metrics['val_recall']
  val_precision = best_epoch_metrics['val_precision']

  # Calculate F1 score, handling potential division by zero
  if (val_precision + val_recall) == 0:
      val_f1 = 0.0
  else:
      val_f1 = 2 * (val_precision * val_recall) / (val_precision + val_recall)

  # 21. Time
  time = round(time, 1) # note I'm rounding to 1 place here

  return {
    "num_hidden_layers": num_hidden_layers,
    "hidden_layer_sizes": hidden_layer_sizes,
    "activation_functions": activation_function_names,
    "epochs": total_epochs,
    "batch_size": batch_size,
    "optimizer": optimizer_name,
    "lr_momentum": [learning_rate_rounded, momentum_rounded],
    "weight_initializer": weight_initializer_name,
    "regularization": regularization_name,
    "dropout_rates": dropout_rates if dropout_rates else "-",
    "train_loss": round(train_loss, 4),
    "train_accuracy": round(train_accuracy, 4),
    "train_recall": round(train_recall, 4),
    "train precision": round(train_precision, 4),
    "train f1": round(train_f1, 4),
    "val_loss": round(val_loss, 4),
    "val_accuracy": round(val_accuracy, 4),
    "val_recall": round(val_recall, 4),
    "val_precision": round(val_precision, 4),
    "val_f1": round(val_f1, 4),
    "training_time": round(time, 1)
}
```

##Model 1: Convolutional Neural Network (CNN) from Scratch

```python
model_1 = Sequential()
model_1.add(Conv2D(64, (3, 3), activation='relu', padding="same", input_shape=(200, 200, 3)))
model_1.add(MaxPooling2D((2, 2), padding = 'same'))
model_1.add(Conv2D(32, (3, 3), activation='relu', padding="same"))
model_1.add(MaxPooling2D((2, 2), padding = 'same'))
model_1.add(Conv2D(32, (3, 3), activation='relu', padding="same"))
model_1.add(MaxPooling2D((2, 2), padding = 'same'))
model_1.add(Flatten())
model_1.add(Dense(100, activation='relu'))
model_1.add(Dense(1, activation='sigmoid')) # Changed activation to sigmoid
opt = SGD(learning_rate=0.01, momentum=0.9)
model_1.compile(optimizer=opt, loss='binary_crossentropy', metrics=metrics_list)
model_1.summary()
```

```python
model_1, history_1, total_epochs_1, total_time_1, best_epoch_metrics_1 = train_with_early_stopping(model_1, X_train, y_train, X_val, y_val, batch_size=64, initial_history=None, patience=10, initial_epochs=0, verbose=1)
```

```python
results_df = pd.DataFrame(columns=new_results_entry(
    model_1,
    history_1,
    total_epochs_1,
    best_epoch_metrics_1,
    total_time_1,
    x_train_data=X_train,
    y_train_data=y_train,
    batch_size=64,
    x_val_data=X_val,
    y_val_data=y_val
).keys())

pd.set_option('display.max_columns', None)
# Display all the columns
```

```python
entry_data_1 = new_results_entry(model_1, history_1, total_epochs_1, best_epoch_metrics_1, total_time_1, X_train, y_train, X_val, y_val, batch_size=64)
results_df.loc[len(results_df)] = pd.Series(entry_data_1, index=results_df.columns)
results_df
```

## Model 3b: Transfer Learning with VGG-16 (Base + FFNN) + Class Weights


```python
# Use VGG-16 base model
base_model = VGG16(weights='imagenet', include_top=False, input_shape=(200, 200, 3))

# Freeze the layers of the base model
for layer in base_model.layers:
    layer.trainable = False

for layer in base_model.layers[-4:]:
    layer.trainable = True

# Add custom classification head on top of the VGG16 base
x = Flatten()(base_model.output)
x = Dense(256, activation='relu')(x)
x = Dropout(0.3)(x)
x = Dense(1, activation='sigmoid')(x) # Output layer for binary classification

# Create the new model
model_3b = Model(inputs=base_model.input, outputs=x)

# Compile the model with standard binary_crossentropy
opt = Adam(learning_rate=1e-4)
model_3b.compile(optimizer=opt, loss='binary_crossentropy', metrics=metrics_list)

model_3b.summary()
```

## Model 4: Transfer Learning with VGG-16 (Base + FFNN + Data Augmentation)

- In most of the real-world case studies, it is challenging to acquire a large number of images and then train CNNs.
- To overcome this problem, one approach we might consider is **Data Augmentation**.
- CNNs have the property of **translational invariance**, which means they can recognise an object even if its appearance shifts translationally in some way. - Taking this attribute into account, we can augment the images using the techniques listed below

    -  Horizontal Flip (should be set to True/False)
    -  Vertical Flip (should be set to True/False)
    -  Height Shift (should be between 0 and 1)
    -  Width Shift (should be between 0 and 1)
    -  Rotation (should be between 0 and 180)
    -  Shear (should be between 0 and 1)
    -  Zoom (should be between 0 and 1) etc.

Remember, **data augmentation should not be used in the validation/test data set**.

```python
datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest',
    preprocessing_function=preprocess_input # Apply VGG16 preprocessing during initialization
)

# Fit the data generator on the training data
datagen.fit(X_train_unnormalized)

# Create a training data generator
train_generator_vgg16_augmented = datagen.flow(
    X_train_unnormalized,
    y_train,
    batch_size=64,
    shuffle=True
)
```

```python
# Use VGG-16 base model
base_model = VGG16(weights='imagenet', include_top=False, input_shape=(200, 200, 3))

# Freeze the layers of the base model
for layer in base_model.layers:
    layer.trainable = False

for layer in base_model.layers[-4:]:
    layer.trainable = True

# Add custom classification head on top of the VGG16 base
x = Flatten()(base_model.output)
x = Dense(256, activation='relu')(x)
x = Dropout(0.3)(x)
x = Dense(1, activation='sigmoid')(x) # Output layer for binary classification

# Create the new model
model_4 = Model(inputs=base_model.input, outputs=x)

# Compile the model with standard binary_crossentropy
opt = Adam(learning_rate=1e-4)
model_4.compile(optimizer=opt, loss='binary_crossentropy', metrics=metrics_list)

model_4.summary()
```

```python
model_4, history_4, total_epochs_4, total_time_4, best_epoch_metrics_4 = train_with_early_stopping(model_4, train_generator_vgg16_augmented, y_train, X_val_vgg16, y_val, batch_size=64, initial_history=None, patience=10, initial_epochs=0, verbose=1)
```

```python
entry_data_4 = new_results_entry(model_4, history_4, total_epochs_4, best_epoch_metrics_4, total_time_4, train_generator_vgg16_augmented, y_train, X_val_vgg16, y_val, batch_size=64)
results_df.loc[len(results_df)] = pd.Series(entry_data_4, index=results_df.columns)
results_df
```

### Visualizing the predictions

```python
plot_loss_recall_accuracy_precision(history_4)
```

```python
plot_confusion_matrix(model_4, X_val_vgg16, y_val)
```

This model (VGG-16 base with dropout and data augmentation) had a training accuracy of around 97.5%, training recall of around 99.2% training precision of around 97.6%, and training f1 of around 98.4%. It also had a validation accuracy of around 97.5%, validation recall of around 98.5%, validation precision of around 98.3%, and validation f1 of around 98.4%. Because the validation recall and precision are approximately equal, we have an approximately equal rate of Type I and Type II errors.

This implies that the model did not necessarily have training metrics being much better than validation metrics, which indicates our model likely did not overfit. This may be because, with data augmentation, we had much more training data available to us.

However, with data augmentation, our model also took nearly 3x the time to train.
